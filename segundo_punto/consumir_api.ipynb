```python
import requests
import json
```

# URL base de la API

```python
#cambiarla http://127.0.0.1:5000 por su dirrecion
BASE_URL = 'http://127.0.0.1:5000/vulnerabilidades'
```

# 1. Obtener y mostrar los primeros 5 registros


```python
response = requests.get(BASE_URL)

if response.status_code == 200:
    vulnerabilidades = response.json()
    print("--- Primeros 5 registros ---")
    for v in vulnerabilidades[:5]:
        print(json.dumps(v, indent=2, ensure_ascii=False))
else:
    print("Error al obtener datos de la vulnerabilidades:", response.status_code)
```

# 2.Agregar un dato a vulnerabilidad

```python
nueva_vuln = {
    "IP": "192.168.1.222",
    "First Detected": "2025-04-10",
    "Last Detected": "2025-04-12",
    "CVE ID": "CVE-2025-9876",
    "Gid": 4,
    "Categoria": "Parche de seguridad",
    "Riesgo": "Medio",
    "Criticidad": "General"
}

post_response = requests.post(BASE_URL, json=nueva_vuln)

if post_response.status_code == 201:
    print("datos agregados exitosamente.")
else:
    print(" Error al agregar datos:", post_response.text)
```
